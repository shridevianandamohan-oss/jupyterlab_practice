```python
def compute_mean(values):
    mean = sum(values) / len(values)
    print(mean)

compute_mean([2, 4, 6, 8])
```

The `compute_mean` function calculates the mean of a list of values. It adds all the values together and divides the total by the number of values in the list.

The mean can be written in two different ways:

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_i
$$

or

$$
\bar{x} = \frac{x_1+x_2+\cdots+x_n}{n}
$$
